# Backtest a symbol (paper, deterministic)

The `BTFW` function runs the bundled walk-forward backtest on the same lake/engine the bots use. No keys required; results are computed locally.

```python
import requests

BASE = "http://127.0.0.1:8795"

bt = requests.get(
    f"{BASE}/api/fn/BTFW",
    params={"symbol": "BTCUSDT", "interval": "1h", "days": 90},
).json()
print("status:", bt.get("status"))
data = bt.get("data") or {}
print("keys:", list(data)[:12])
```

```python
# Risk parity weights across a mixed universe (RPAR)
rpar = requests.get(
    f"{BASE}/api/fn/RPAR",
    params={"symbols": "AAPL,MSFT,BTCUSDT,EURUSD,GC=F", "live_risk": "true"},
).json()
meta = rpar.get("metadata") or {}
print("fallback:", meta.get("fallback"), meta.get("fallback_reason"))
for row in (rpar.get("data") or {}).get("rows", [])[:5]:
    print(row)
```
