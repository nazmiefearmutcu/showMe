# showMe Quickstart

The packaged sidecar exposes the engine through plain HTTP routes; every pane in the terminal is a thin client of the same surface.

- `GET /api/bars` - OHLCV bars (Binance for crypto, Yahoo for the rest)
- `GET /api/fn/<CODE>` - any of the 157 registered functions
- `GET /api/quote/<SYMBOL>` - quote snapshots

This notebook is a starting point for your own research; run it against a locally running sidecar (default port 8795).

```python
import requests

BASE = "http://127.0.0.1:8795"

bars = requests.get(
    f"{BASE}/api/bars",
    params={"symbol": "AAPL", "interval": "1D", "limit": 30},
).json()
print(bars["source"], len(bars["bars"]), "bars; last close:", bars["bars"][-1]["c"])
```

```python
# Engine functions return their pane payloads verbatim
des = requests.get(f"{BASE}/api/fn/DES", params={"symbol": "MSFT"}).json()
print(des["data"].get("name"), "| market cap:", des["data"].get("market_cap"))
```

```python
# World-events tracker: upcoming high-impact calendar rows
meet = requests.get(f"{BASE}/api/fn/MEET", params={"impact": "high"}).json()
for row in meet["data"]["upcoming"][:5]:
    print(row["when_utc"], row.get("country_names"), "-", row["title"])
```
